#### Ejercicio 1

Implementar un sampler simple utilizando la muestra piano.wav. Mapear las teclas zxc.. . a una
octava del piano y qwe... a la octava superior. A partir de muestra de una nota el resto de notas
pueden obtenerse variando la velocidad de reproducción de esa muestra de acuerdo a estas propor-
ciones. Por ejemplo, asumiento que la nota dada es un C, el resto de notas pueden obtenerse con
las siguientes proporciones:

Nota  C  D   E   F   G   A    B  c . . .
Frec. 1 9/8 5/4 4/3 3/2 5/3 15/8 2 . . .

Es decir, para obtener un D (re) reproduciríamos el sample (virtualmente) a 1.12 de velocidad.
Observemos que dado un array de NumPy, si generamos otro que contenga solo los samples en
posición par (de la mitad de tamaño), obtenemos un sonido una octava por arriba. Y al contrario,
si generamos un array del doble de tamaño añadiendo muestras intermedias (como media de la
anterior y la superior de una dada) obtenemos una octava menos. Esta idea de interpolación puede
generalizarse para obtener cualquier nota, utitilizando las proporciones dadas. Esto, a us vez puede
hacerse de manera explícida o utilizando primitivas de numPy os scipy (como resample).
2. Implementar un sampler que permita cargar muestras de audio preparadas para procesar en tiempo
real. En particular, para una nota de piano como la de la primera pista de abajo, podemos utilizar
dos fragmentos. El primero con la parte de ataque y decay, y el segundo con una zona loopeable
para construir el sustain (la última parte de la muestra inicial la podemos descartar).
El fragmento de sustain debe elegirse adecuadamente para poder construir el loop. Sobre ese frag-
mento hay que linealizar el volumen y asegurarse de que suena bien cíclicamente.
Estos dos fragmentos serán los que cargue el sampler. El mensaje noteOn activará la generación de
una nota: se reproduce la región de ataque-decay y luego se reproduce en loop la región de sustain
1Afinación justa, también puede hacer con temperada, tal como se hizo en la hoja anterior


```python
from files.consts import *
from files.speedSampler import Sampler
from importlib import reload
from tkinter import *
import sounddevice as sd
import numpy as np

speeds = [1, 9/8, 5/4, 4/3, 3/2, 5/3, 15/8, 2, 18/8, 10/4, 8/3, 3, 10/3, 30/8]
sampler = Sampler(1)

def on_key_pressed(event):
    global sampler;
    key = event.char

    if key in teclas:
        freqIndex = teclas.index(key)
        speed = speeds[freqIndex]
        sampler = Sampler(speed)

def callback(outdata, frames, time, status):
    global sampler
    
    chunk = np.zeros(CHUNK)
    if sampler != None: chunk = sampler.next()

    chunk = np.reshape(chunk,(chunk.shape[0],1))
    outdata[:CHUNK] = chunk

stream = sd.OutputStream(samplerate=SRATE, channels=1,
    callback=callback, blocksize=CHUNK)
stream.start() 

tkinter=Tk()

tkinter.bind("<Key>",on_key_pressed)
tkinter.mainloop()

stream.close()
```
