```python
from files.consts import *
from files.speedSampler import Sampler
from importlib import reload
from tkinter import *
import sounddevice as sd
import numpy as np

speeds = [1, 9/8, 5/4, 4/3, 3/2, 5/3, 15/8, 2, 18/8, 10/4, 8/3, 3, 10/3, 30/8]
sampler = Sampler(1)

def on_key_pressed(event):
    global sampler;
    key = event.char

    if key in teclas:
        freqIndex = teclas.index(key)
        speed = speeds[freqIndex]
        sampler = Sampler(speed)

def callback(outdata, frames, time, status):
    global sampler
    
    chunk = np.zeros(CHUNK)
    if sampler != None: chunk = sampler.next()

    chunk = np.reshape(chunk,(chunk.shape[0],1))
    outdata[:CHUNK] = chunk

stream = sd.OutputStream(samplerate=SRATE, channels=1,
    callback=callback, blocksize=CHUNK)
stream.start() 

tkinter=Tk()

tkinter.bind("<Key>",on_key_pressed)
tkinter.mainloop()

stream.close()
```
